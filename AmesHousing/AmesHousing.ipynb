```python
import pandas as pd

df = pd.read_excel("AmesHousing.xlsx", sheet_name="AmesHousing")

kolommen = ["Overall Qual", "Gr Liv Area", "Neighborhood", "SalePrice"]
data = df[kolommen].copy()

data_encoded = pd.get_dummies(data, columns=["Neighborhood"], drop_first=False)

# Horizontaal en verticaal splitsen
n_rows, n_cols = data_encoded.shape
mid_row = n_rows // 2
mid_col = n_cols // 2

stuk1 = data_encoded.iloc[:mid_row, :mid_col]
stuk2 = data_encoded.iloc[:mid_row, mid_col:]
stuk3 = data_encoded.iloc[mid_row:, :mid_col]
stuk4 = data_encoded.iloc[mid_row:, mid_col:]

print("Originele dataset:", data_encoded.shape)
print("Stuk1:", stuk1.shape)
print("Stuk2:", stuk2.shape)
print("Stuk3:", stuk3.shape)
print("Stuk4:", stuk4.shape)
```

Opdracht 5

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

#Train set
X_train = stuk1.drop("SalePrice", axis=1)
y_train = stuk1["SalePrice"]

#Test set
X_test = stuk3.drop("SalePrice", axis=1)
y_test = stuk3["SalePrice"]

#Model met hyperparameters
model = LinearRegression(
    fit_intercept=True,
    positive=True
)

#Trainen
model.fit(X_train, y_train)

#Voorspellen
y_pred = model.predict(X_test)

#Evaluatie
print("R² score:", r2_score(y_test, y_pred))
```

Opdracht 6

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

# R²
r2 = r2_score(y_test, y_pred)

# MAE
mae = mean_absolute_error(y_test, y_pred)

# MSE
mse = mean_squared_error(y_test, y_pred)

# RMSE
rmse = np.sqrt(mse)

print("R² score:", r2)
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
```

```python
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

df = pd.read_excel("AmesHousing.xlsx", sheet_name="AmesHousing")

kolommen = [
    "Overall Qual",
    "Gr Liv Area",
    "Neighborhood",
    "SalePrice"
]

data = df[kolommen].copy()
data = pd.get_dummies(data, columns=["Neighborhood"], drop_first=False)
data = data.dropna()

n_rows, n_cols = data.shape
mid_row = n_rows // 2
mid_col = n_cols // 2

stuk1 = data.iloc[:mid_row, :mid_col]
stuk3 = data.iloc[mid_row:, :mid_col]

X_train = stuk1.drop("SalePrice", axis=1)
y_train = stuk1["SalePrice"]

X_test = stuk3.drop("SalePrice", axis=1)
y_test = stuk3["SalePrice"]

model = SGDRegressor(
    max_iter=1000,      # lijkt op epochs
    eta0=0.000001,      # learning rate
    learning_rate="constant",
    random_state=42
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("R²:", r2)
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
```
