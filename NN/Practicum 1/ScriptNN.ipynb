# Wine Quality Neural Network Experimenten

In deze notebook worden meerdere neural-network instellingen getest voor de wine-quality dataset. De target is binair gemaakt: `quality >= 6` betekent goede wijn, anders minder goede wijn.

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import pandas as pd

nn_dir = Path.cwd()
if not (nn_dir / "run_wine_experiments.py").exists():
    nn_dir = Path.cwd() / "NN"
sys.path.insert(0, str(nn_dir.resolve()))

from run_wine_experiments import run_experiments

OUTPUT_DIR = nn_dir / "experiment_results"
```

## Experimenten uitvoeren

De onderstaande cell traint 30 modellen met verschillende instellingen, zoals aantal hidden layers, hidden nodes, epochs, learning rate, batch size, activatie en dropout. De resultaten worden opgeslagen als CSV en Excel.

```python
results = run_experiments()
results.head(10)
```
